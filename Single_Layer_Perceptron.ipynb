```python
import numpy as np
```

```python
def step_function(x):
    if x >= 0:
        return 1
    else:
        return 0
```

```python
def train_perceptron(X, y, learning_rate=0.1, epochs=10):

    weights = np.zeros(X.shape[1])
    bias = 0

    for epoch in range(epochs):

        total_error = 0

        for i in range(len(X)):

            weighted_sum = np.dot(X[i], weights) + bias

            prediction = step_function(weighted_sum)

            error = y[i] - prediction

            total_error = total_error + abs(error)

            weights = weights + learning_rate * error * X[i]

            bias = bias + learning_rate * error

        print("Epoch:", epoch + 1, "Error:", total_error)

    return weights, bias
```

```python
def predict(X, weights, bias):

    output = []

    for x in X:

        weighted_sum = np.dot(x, weights) + bias

        prediction = step_function(weighted_sum)

        output.append(prediction)

    return output
```

```python
X_AND = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_AND = np.array([0, 0, 0, 1])
```

```python
weights_AND, bias_AND = train_perceptron(X_AND, y_AND)
```

```python
print("AND Gate Output:")

result = predict(X_AND, weights_AND, bias_AND)

print(result)
```

```python
X_OR = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y_OR = np.array([0, 1, 1, 1])
```

```python
weights_OR, bias_OR = train_perceptron(X_OR, y_OR)
```

```python
print("OR Gate Output:")

result = predict(X_OR, weights_OR, bias_OR)

print(result)
```

```python
X_NOT = np.array([
    [0],
    [1]
])

y_NOT = np.array([1, 0])
```

```python
weights_NOT, bias_NOT = train_perceptron(X_NOT, y_NOT)
```

```python
print("NOT Gate Output:")

result = predict(X_NOT, weights_NOT, bias_NOT)

print(result)
```
